##Punto 13: expresión de la salida en el tiempo para una configuración en lazo cerrado del sistema, en función de R, L y C

En este ejercicio se encontró la función de transferencia en lazo cerrado para el sistema,
en función de R, L y C, para las siguientes entradas:

- i) impulso
- ii) escalón unitario
- iv) rampa.

Se Comprobo los resultados en simulación sobre Python (utililzando sympy) y se grafico la señal de salida para cada caso: [texto del vínculo](https://)

###Código de Simulación en Python para este ejercicio

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# -- PASO 1: DEFINICIÓN SIMBÓLICA --
# Definimos las variables simbólicas que usaremos
s, t, R, L, C = sp.symbols('s t R L C', real=True, positive=True)

# Definimos las funciones de transferencia en LAZO ABIERTO G(s)
# Sistema 1: RLC Serie (Salida en Capacitor)
G1 = 1 / (L*C*s**2 + R*C*s + 1)

# Sistema 2: RLC Paralelo (Salida en Inductor)
G2 = 1 / (L*C*s**2 + (L/R)*s + 1)

# Calculamos las funciones de transferencia en LAZO CERRADO T(s) = G(s)/(1+G(s))
T1 = sp.cancel(G1 / (1 + G1))
T2 = sp.cancel(G2 / (1 + G2))

print("="*50)
print("FUNCIONES DE TRANSFERENCIA EN LAZO CERRADO")
print("="*50)
display(sp.Eq(sp.Symbol('T1(s)'), T1))
print("\n")
display(sp.Eq(sp.Symbol('T2(s)'), T2))

# Definimos las entradas R(s)
R_impulso = 1
R_escalon = 1/s
R_rampa = 1/s**2

# Creamos un diccionario para facilitar el manejo
entradas = {
    'Impulso': R_impulso,
    'Escalón': R_escalon,
    'Rampa': R_rampa
}

# -- PASO 2: CÁLCULO DE LAS EXPRESIONES EN EL TIEMPO --
# Esta función calcula la transformada inversa de Laplace
def obtener_salida_tiempo(T_s, R_s):
    C_s = T_s * R_s
    # Usamos sympy para la transformada inversa
    c_t = sp.inverse_laplace_transform(C_s, s, t)
    return sp.simplify(c_t)

# Calculamos y mostramos las 6 expresiones de salida en el tiempo c(t)
print("\n" + "="*50)
print("EXPRESIONES DE SALIDA EN EL TIEMPO c(t)")
print("="*50)

# Para Sistema 1
print("\n--- Sistema 1: RLC Serie ---")
for nombre, R_s in entradas.items():
    c1_t = obtener_salida_tiempo(T1, R_s)
    display(sp.Eq(sp.Symbol(f'c1_{nombre.lower()}(t)'), c1_t))
    print("\n") # Add spacing here

# Para Sistema 2
print("\n--- Sistema 2: RLC Paralelo ---")
for nombre, R_s in entradas.items():
    c2_t = obtener_salida_tiempo(T2, R_s)
    display(sp.Eq(sp.Symbol(f'c2_{nombre.lower()}(t)'), c2_t))
    print("\n") # Add spacing here


# -- PASO 3: SIMULACIÓN NUMÉRICA Y GRÁFICAS --

# Definimos valores numéricos para R, L, C para poder graficar
# Puedes cambiar estos valores para ver cómo se comporta el sistema
valores_numericos_1 = {R: 1.0, L: 1.0, C: 1.0} # Sistema 1: Subamortiguado
valores_numericos_2 = {R: 2.0, L: 1.0, C: 2.0} # Sistema 2: Subamortiguado

# Creamos un vector de tiempo para la simulación
t_vals = np.linspace(0, 25, 1000)

# Definimos una lista de colores para las gráficas
colores = ['blue', 'green', 'purple']

# Función para preparar y generar las gráficas
def graficar_respuestas(sistema_num, T_s, valores, titulo):
    fig, axs = plt.subplots(1, 3, figsize=(20, 5))
    fig.suptitle(f'Respuestas en Lazo Cerrado del Sistema {sistema_num}: {titulo}', fontsize=16)

    for i, (nombre, R_s) in enumerate(entradas.items()):
        # Obtener la expresion c(t) y sustituir los valores numéricos
        c_t_expr = obtener_salida_tiempo(T_s, R_s)
        c_t_numerica = c_t_expr.subs(valores)

        # Convertir la expresión simbólica de sympy a una función numérica para graficar rápido
        func_numpy = sp.lambdify(t, c_t_numerica, 'numpy')

        # Calcular los valores de y para el vector de tiempo
        y_vals = func_numpy(t_vals)

        # Graficar
        axs[i].plot(t_vals, y_vals, label=f'Salida $c(t)$', lw=2, color=colores[i % len(colores)])
        axs[i].set_title(f'Respuesta a una entrada "{nombre}"')
        axs[i].set_xlabel('Tiempo (s)')
        axs[i].set_ylabel('Amplitud')
        axs[i].grid(True)
        axs[i].legend()

        # Graficar la entrada de referencia para escalón y rampa
        if nombre == 'Escalón':
            axs[i].plot(t_vals, np.ones_like(t_vals), 'r--', label='Entrada (Escalón)')
            axs[i].legend()
        elif nombre == 'Rampa':
            axs[i].plot(t_vals, t_vals, 'r--', label='Entrada (Rampa)')
            axs[i].legend()

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

# Generar las gráficas para ambos sistemas
graficar_respuestas(1, T1, valores_numericos_1, "RLC Serie (Salida en Capacitor)")
graficar_respuestas(2, T2, valores_numericos_2, "RLC Paralelo (Salida en Inductor)")
```
