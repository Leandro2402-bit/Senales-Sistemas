# Análisis de Series y Transformadas de Fourier

## 1. Serie de Fourier

### Serie Exponencial Compleja
Para una señal periódica x(t) con período T₀:

```
x(t) = Σ(n=-∞ to ∞) cₙ e^(jnω₀t)
```

Donde:
- ω₀ = 2π/T₀ (frecuencia fundamental)
- cₙ = (1/T₀) ∫[T₀] x(t) e^(-jnω₀t) dt (coeficientes de Fourier)

### Serie Trigonométrica
```
x(t) = a₀/2 + Σ(n=1 to ∞) [aₙcos(nω₀t) + bₙsen(nω₀t)]
```

Donde:
- a₀ = (2/T₀) ∫[T₀] x(t) dt
- aₙ = (2/T₀) ∫[T₀] x(t) cos(nω₀t) dt
- bₙ = (2/T₀) ∫[T₀] x(t) sen(nω₀t) dt

### Serie Compacta
```
x(t) = A₀ + Σ(n=1 to ∞) Aₙcos(nω₀t + φₙ)
```

Donde:
- A₀ = a₀/2
- Aₙ = √(aₙ² + bₙ²)
- φₙ = -arctan(bₙ/aₙ)

## 2. Transformada de Fourier

### Transformada Continua
```
X(jω) = ∫(-∞ to ∞) x(t) e^(-jωt) dt    (Transformada directa)
x(t) = (1/2π) ∫(-∞ to ∞) X(jω) e^(jωt) dω    (Transformada inversa)
```

## 3. Transformada de Fourier en Tiempo Discreto (DTFT)

Para señales discretas x[n]:
```
X(e^(jΩ)) = Σ(n=-∞ to ∞) x[n] e^(-jΩn)    (DTFT directa)
x[n] = (1/2π) ∫(-π to π) X(e^(jΩ)) e^(jΩn) dΩ    (DTFT inversa)
```

Donde Ω es la frecuencia angular normalizada.

## 4. Transformada Discreta de Fourier (DFT)

Para secuencias finitas de N puntos:
```
X[k] = Σ(n=0 to N-1) x[n] e^(-j2πkn/N)    (DFT directa)
x[n] = (1/N) Σ(k=0 to N-1) X[k] e^(j2πkn/N)    (DFT inversa)
```

## 5. Diferencias entre Espectros

### Espectro Continuo
- Se obtiene de la Transformada de Fourier continua
- Representa señales aperiódicas
- Densidad espectral de potencia/energía
- Dominio de frecuencia continuo

### Espectro Discreto
- Se obtiene de la Serie de Fourier
- Representa señales periódicas
- Líneas espectrales en frecuencias armónicas
- Amplitudes discretas en múltiplos de la frecuencia fundamental

## 6. Combinaciones entre Señal y Espectro

| Señal | Espectro | Ejemplo |
|-------|----------|---------|
| Continua y Aperiódica | Continuo y Aperiódico | Pulso rectangular |
| Continua y Periódica | Discreto y Aperiódico | Onda sinusoidal |
| Discreta y Aperiódica | Continuo y Periódico | Secuencia finita |
| Discreta y Periódica | Discreto y Periódico | Secuencia periódica muestreada |

## 7. Fast Fourier Transform (FFT)

### ¿Qué es la FFT?
La FFT es un algoritmo eficiente para calcular la DFT. Reduce la complejidad computacional de O(N²) a O(N log N).

### Algoritmo FFT (Decimación en Tiempo)
1. **Divide**: Separa la secuencia en subsecuencias pares e impares
2. **Conquista**: Calcula recursivamente la DFT de cada subsecuencia
3. **Combina**: Combina los resultados usando factores de rotación (twiddle factors)

```python
# código básico
def fft(x):
    N = len(x)
    if N <= 1:
        return x
    
    # Dividir en pares e impares
    even = fft(x[0::2])
    odd = fft(x[1::2])
    
    # Factores de rotación
    T = [exp(-2j*pi*k/N) * odd[k] for k in range(N//2)]
    
    # Combinar
    return [even[k] + T[k] for k in range(N//2)] + \
           [even[k] - T[k] for k in range(N//2)]
```

### Utilidad de la FFT
- **Procesamiento de señales**: Filtrado, análisis espectral
- **Convolución rápida**: Multiplicación en dominio de frecuencia
- **Compresión de datos**: JPEG, MP3
- **Resolución de ecuaciones diferenciales**: Métodos espectrales

## 8. Costo Computacional

### DFT Directa: O(N²)
- N² multiplicaciones complejas
- Para N = 1024: ~1,000,000 operaciones

### FFT: O(N log₂ N)
- Para N = 1024: ~10,000 operaciones
- **Mejora de ~100x para N = 1024**

### Comparación práctica:
| N | DFT (N²) | FFT (N log₂ N) | Mejora |
|---|----------|----------------|---------|
| 64 | 4,096 | 384 | 10.7x |
| 256 | 65,536 | 2,048 | 32x |
| 1024 | 1,048,576 | 10,240 | 102x |
| 4096 | 16,777,216 | 49,152 | 341x |

La FFT es esencial para aplicaciones en tiempo real debido a esta eficiencia computacional significativa.
