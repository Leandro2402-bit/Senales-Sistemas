### 📊 Animación de Constelación 16-QAM

Este bloque de código simula la transmisión de símbolos digitales mediante **modulación 16-QAM** y los visualiza uno por uno sobre el plano I/Q en forma de **animación**.

#### 🔹 ¿Qué hace el código?

- Define los **niveles de amplitud** de la constelación 16-QAM: {−3, −1, +1, +3}.
- Genera aleatoriamente una secuencia de **símbolos QAM** que representan combinaciones de I (en fase) y Q (en cuadratura).
- Muestra una **constelación base estática** (todos los posibles símbolos).
- Destaca cada símbolo recibido de forma **animada** en color rojo, simulando su transmisión uno a uno.

#### 🎯 ¿Para qué sirve?

- Visualizar cómo los símbolos se transmiten en el plano I/Q.
- Ilustrar el funcionamiento de una constelación digital.
- Usar como recurso educativo en el video final para explicar la modulación QAM.

---

Cada punto rojo en la animación representa un **símbolo 16-QAM** que codifica 4 bits. Esta técnica es ampliamente usada en comunicaciones WiFi y 5G por su eficiencia para transmitir grandes volúmenes de datos.


```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation

# ----- Parámetros -----
np.random.seed(42)
levels = [-3, -1, 1, 3]  # 16-QAM
symbols = np.array([(i, q) for i in levels for q in levels])  # Todos los puntos posibles
stream = np.random.choice(len(symbols), size=100, replace=True)  # Secuencia de símbolos

# ----- Crear figura -----
fig, ax = plt.subplots(figsize=(6, 6))
ax.set_xlim(-4, 4)
ax.set_ylim(-4, 4)
ax.set_xticks(levels)
ax.set_yticks(levels)
ax.set_title('Animación de Constelación 16-QAM')
ax.set_xlabel('I (In-phase)')
ax.set_ylabel('Q (Quadrature)')
ax.grid(True)
ax.axhline(0, color='gray', linewidth=0.5)
ax.axvline(0, color='gray', linewidth=0.5)

# Puntos base
ax.scatter(symbols[:,0], symbols[:,1], color='lightblue', alpha=0.3, label='Constelación base')

# Punto activo
point, = ax.plot([], [], 'ro', label='Símbolo recibido', markersize=8)

# ----- Función de animación -----
def update(frame):
    i, q = symbols[stream[frame]]
    point.set_data([i], [q]) # Pass i and q as lists
    return point,

ani = animation.FuncAnimation(fig, update, frames=len(stream), interval=150, blit=True)

# Guardar como GIF
ani.save('constelacion_qam.gif', writer='pillow')
plt.close()
```
